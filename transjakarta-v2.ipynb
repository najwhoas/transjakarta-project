## Cohort Analysis

Cohort analysis adalah metode analisis yang mengelompokkan pengguna atau data berdasarkan kesamaan waktu kejadian tertentu, seperti tanggal pendaftaran atau pembelian pertama, untuk melacak perilaku atau retensi mereka dari waktu ke waktu.

```python
import datetime as dt
```

```python
data['current_week'] = pd.to_datetime(data['tapOutTime']).dt.to_period('W').apply(lambda r: r.start_time)
```

```python
data["cohort_week"] = data.groupby("payCardID")["current_week"].transform("min")
```

```python
def cohort_index(df, col1, col2):
    current_week = pd.to_datetime(df[col1]).dt.isocalendar().week
    cohort_week = pd.to_datetime(df[col2]).dt.isocalendar().week
    return current_week - cohort_week

data['cohort_index'] = cohort_index(data, "current_week", "cohort_week")
```

```python
def weeks_diff(df, col1, col2):
    current_week = pd.to_datetime(df[col1]).dt.isocalendar().week
    cohort_week = pd.to_datetime(df[col2]).dt.isocalendar().week
    return current_week - cohort_week

data['weeks_diff'] = weeks_diff(data, "current_week", "cohort_week")
```

```python
cohort_counts = data.groupby(['cohort_week', 'cohort_index'])['payCardID'].nunique()
```

```python
cohort_counts_df = cohort_counts.to_frame().rename(columns={'payCardID':'users'}) \
                                .sort_values(by=['cohort_week'])

cohort_counts_df['retention_rate'] = cohort_counts_df['users'] / cohort_counts_df.groupby(["cohort_week"])["users"].transform("max")
```

```python
cohorts_pivot = cohort_counts_df.pivot_table(index='cohort_week', columns='cohort_index', values='retention_rate')
```

Grafik menunjukkan bahwa retensi cenderung menurun setiap minggu, di mana setiap kohort mengalami penurunan persentase pengguna aktif di minggu-minggu setelah mereka bergabung. Kohort yang lebih awal (seperti 2023-03-27) cenderung memiliki retensi lebih tinggi pada minggu-minggu berikutnya dibandingkan dengan kohort yang bergabung lebih belakangan.

```python
plt.figure(figsize = (12, 10))
sns.heatmap(data = cohorts_pivot,annot = True, fmt = '.2%', vmin = 0.0, vmax = 0.1, cmap = 'BuGn')
```

## RFM Analysis
Analisis RFM (Recency, Frequency, Monetary) adalah metode analisis perilaku pelanggan yang digunakan untuk mengelompokkan pelanggan berdasarkan tiga variabel:

*   Recency: Kapan terakhir kali pelanggan melakukan transaksi
* Frequency: Seberapa sering pelanggan melakukan transaksi
* Monetary: Seberapa besar pendapatan yang berasal dari pelanggan tersebut




```python
# Konversi tapOutTime ke format datetime
data['tapOutTime'] = pd.to_datetime(data['tapOutTime'])
```

```python
# Menyiapkan data untuk RFM Analysis
# Recency: Waktu dari transaksi terakhir
recency_df = data.groupby('payCardName').tapOutTime.max().reset_index()
recency_df['Recency'] = (recency_df['tapOutTime'].max() - recency_df['tapOutTime']).dt.days

# Frequency: Jumlah perjalanan yang dilakukan oleh setiap penumpang
frequency_df = data.groupby('payCardName').tapOutTime.count().reset_index()
frequency_df.columns = ['payCardName', 'Frequency']

# Monetary: Total biaya yang dikeluarkan setiap penumpang
monetary_df = data.groupby('payCardName').payAmount.sum().reset_index()
monetary_df.columns = ['payCardName', 'Monetary']
```

```python
# Menggabungkan RFM
rfm_df = recency_df.merge(frequency_df, on='payCardName').merge(monetary_df, on='payCardName')
print(rfm_df.head())
```

Mayoritas penumpang memiliki transaksi yang sangat baru (0–5 hari yang lalu), yang menunjukkan keterlibatan yang tinggi. Terdapat puncak pada kategori 0–1 hari, dengan sekitar 700 penumpang, menunjukkan bahwa sebagian besar pelanggan melakukan transaksi baru-baru ini.

```python
# Distribusi Recency
sns.histplot(rfm_df['Recency'], bins=20, kde=True)
plt.title("Distribusi Recency")
plt.xlabel("Hari Sejak Transaksi Terakhir")
plt.ylabel("Jumlah Penumpang")
plt.show()
```

Menunjukkan jumlah perjalanan yang dilakukan penumpang. Distribusi ini juga condong, dengan mayoritas penumpang hanya melakukan sedikit perjalanan dan sejumlah kecil bepergian lebih sering.

```python
# Distribusi Frequency
sns.histplot(rfm_df['Frequency'], bins=20, kde=True)
plt.title("Distribusi Frequency")
plt.xlabel("Frekuensi Perjalanan")
plt.ylabel("Jumlah Penumpang")
plt.show()
```

Menunjukkan total pengeluaran penumpang, dengan mayoritas penumpang memiliki pengeluaran rendah dan beberapa di antaranya dengan pengeluaran tinggi sebagai outlier. Distribusi yang condong seperti ini umum terjadi, menunjukkan bahwa banyak penumpang yang hanya sesekali menggunakan layanan, sementara sejumlah kecil sering bepergian dengan pengeluaran lebih besar.

```python
# Distribusi Monetary
sns.histplot(rfm_df['Monetary'], bins=20, kde=True)
plt.title("Distribusi Monetary")
plt.xlabel("Total Biaya (Rp)")
plt.ylabel("Jumlah Penumpang")
plt.show()
```

## CLV Analysis

Customer Lifetime Value (CLV) analysis adalah metode untuk mengukur total nilai pendapatan yang diharapkan dari seorang pelanggan selama hubungan mereka dengan bisnis. CLV membantu perusahaan memahami nilai jangka panjang setiap pelanggan dan mengoptimalkan strategi pemasaran, retensi, dan layanan pelanggan.

```python
# ---- 3. Menghitung CLV ---- #
# Average Transaction Value (rata-rata pembayaran per transaksi)
rfm_df['AverageTransactionValue'] = rfm_df['Monetary'] / rfm_df['Frequency']
```

```python
# Ensure transaction dates are in datetime format
data['transaction_date'] = pd.to_datetime(data['tapOutTime'])

# Calculate the first and last transaction date per customer
rfm_df['FirstTransaction'] = data.groupby('payCardID')['transaction_date'].transform('min')
rfm_df['LastTransaction'] = data.groupby('payCardID')['transaction_date'].transform('max')

# Calculate lifespan (in days) for each customer
rfm_df['LifespanDays'] = (rfm_df['LastTransaction'] - rfm_df['FirstTransaction']).dt.days

# Convert lifespan to weeks
rfm_df['LifespanWeeks'] = rfm_df['LifespanDays'] / 7

# Calculate the average lifespan in weeks
average_lifespan_weeks = rfm_df['LifespanWeeks'].mean()
```

```python
# Now you can use this average in your CLV calculation
rfm_df['CLV'] = rfm_df['AverageTransactionValue'] * rfm_df['Frequency'] * average_lifespan_weeks
print(rfm_df[['payCardName', 'Recency', 'Frequency', 'Monetary', 'AverageTransactionValue', 'CLV']].head())
```
